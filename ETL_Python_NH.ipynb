```python
#import de bibliotecas necessárias

import requests as req
import pandas as pd
from requests.exceptions import HTTPError

#definição de URL e caminho onde a tabela fato será salva

url = 'http://transparencia.al.gov.br/patrimonio/json-patrimonio-imobiliario/?limit=100&offset=0'
caminho_windows = r'C:\Users\Paulo\Desktop\Projeto N.H\fato.csv'
```

```python
#Requisição GET na API e transformação dos dados em DataFrame

requisicao = req.get(url)
try:

    requisicao.raise_for_status()

    dados = requisicao.json()
    df = pd.DataFrame(dados["rows"])
    print('Sucesso')

except HTTPError as htterr:
     print('Erro HTTP:', htterr)
except Exception as err:
     print('Erro normal', err)
```

```python
#Verificar sucesso da importação em outra célula para evitar requisições repetidas sem necessidade

df
```

```python
#Salvando tabela fato

df.to_csv(r'C:\Users\Paulo\Desktop\Projeto N.H\fato.csv', index=False)
```

```python
#Lendo tabela fato para extrair dimensões

df_fato = pd.read_csv(caminho_windows)
```

```python
#Código GPT para verificação de duplicatas

df_fato[df_fato['numero_dossie'].duplicated(keep=False)] \
    .sort_values('numero_dossie')
```

```python
#Código GPT para extração de dimensões da tabela fato

dim_municipio = df_fato[['municipio']].drop_duplicates().reset_index(drop=True)

dim_municipio['id_municipio'] = dim_municipio.index + 1

dim_municipio = dim_municipio[['id_municipio', 'municipio']]

df_fato = df_fato.merge(dim_municipio, on='municipio', how='left')

dim_municipio.to_csv(r'C:\Users\Paulo\Desktop\Projeto N.H\dimensoes\dim_municipio.csv')
```

```python
dim_situacao = df_fato[['situacao_ocupacao']].drop_duplicates().reset_index(drop=True)

dim_situacao['id_situacao'] = dim_situacao.index + 1

dim_situacao = dim_situacao[['id_situacao', 'situacao_ocupacao']]

df_fato = df_fato.merge(dim_situacao, on='situacao_ocupacao', how='left')

dim_situacao.to_csv(r'C:\Users\Paulo\Desktop\Projeto N.H\dimensoes\dim_situacao.csv')
```

```python
dim_bairro = df_fato[['bairro']].drop_duplicates().reset_index(drop=True)

dim_bairro['id_bairro'] = dim_bairro.index + 1

dim_bairro = dim_bairro[['id_bairro', 'bairro']]

df_fato = df_fato.merge(dim_bairro, on='bairro', how='left')

dim_bairro.to_csv(r'C:\Users\Paulo\Desktop\Projeto N.H\dimensoes\dim_bairro.csv')
```

```python
dim_denominacao = df_fato[['denominacao']].drop_duplicates().reset_index(drop=True)

dim_denominacao['id_denominacao'] = dim_denominacao.index + 1

dim_denominacao = dim_denominacao[['id_denominacao', 'denominacao']]

df_fato = df_fato.merge(dim_denominacao, on='denominacao', how='left')

dim_denominacao.to_csv(r'C:\Users\Paulo\Desktop\Projeto N.H\dimensoes\dim_denominacao.csv')
```
